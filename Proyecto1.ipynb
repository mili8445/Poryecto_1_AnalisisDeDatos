```python
#importamos la biblioteca de pandas y abreviamos a pd

import pandas as pd
#importamos numpy
import numpy as np 
```

```python
#Traemos el archivo para leerlo
ventas = pd.read_csv('ventass.csv') 

#usamos shape para saber cuantas filas y columnas tiene. (La dimension) 
print(ventas.shape)

#usamos head para ver las filas y si son muchas columnas muestra las primeras 5 del data.frame para ver que informacion tenemos (un pantallazo)
print(ventas.head())

#para ver la estructura y los tipos de datos utilizamos el metodo info
print(ventas.info())

```

```python
#para ver los valores nulos
print(ventas.isnull().sum())
```

```python
#buscamos que variables numericas hay
#utilizamos el metodo select_dtypes(include=['number']).columns para filtrar y que queden únicamente las columnas numéricas (int, float)
columnas_numericas = ventas.select_dtypes(include='number').columns
print(list(columnas_numericas))
```

```python
#buscamos que variables categoricas hay (str)
#utilizamos el metodo select_dtypes(include=['object', 'category']).columns para filtrar y que queden únicamente las columnas categoricas
columnas_categoricas = ventas.select_dtypes(include=['object', 'category']).columns
print(columnas_categoricas)
```

```python
#elegimos las 3 columnas para las variables numericas 
variable_n1 = ventas["Precio_Unitario"]
variable_n2 = ventas["Venta_Total"]
variable_n3 = ventas["Costo_Total"]

#elegimos las 3 columnas para las variables categoricas 
variable_c1 = ventas["Producto"]
variable_c2 = ventas["Categoria"]
variable_c3 = ventas["Metodo_Pago"]
```

```python
#Convertimos dos columnas a arrays de NumPy
variable_n1 = ventas["Precio_Unitario"].to_numpy(dtype=float)
variable_n2 = ventas["Venta_Total"].to_numpy() 
variable_c1 = ventas["Producto"].to_numpy()
#y lo podemos comprobar con 
print(type(variable_n1))
```

```python
#Calcular: media, mediana, desviación estándar, percentiles (implementar una función propia que automatice esta parte del análisis)
#Media calcula el promedio de todos los valores (sirve para conocer el punto medio real)
#Mediana mide la tendencia central dividiendo los datos en dos partes iguales
#Desviacion estandar mide qué tanto se alejan o dispersan los datos con respecto al promedio
#Percentiles sirve para saber qué posición ocupa un dato en relación con todos los demás


#creamos nuestra funcion 

def calcular(array, nombre_variable="calculo"):
    media = np.mean(array)
    mediana = np.median(array)
    desviacion_estandar = np.std(array)
    percentil25 = np.percentile(array, 25)
    percentil50 = np.percentile(array, 50)
    percentil75 = np.percentile(array, 75)

    print(f"Nombre de variable: {nombre_variable}")
    print(f"Media: {media:.2f}")
    print(f"Mediana: {mediana:.2f}")
    print(f"Desviación Estándar: {desviacion_estandar:.2f}")
    print(f"Percentil 25 (Q1): {percentil25:.2f}")
    print(f"Percentil 50 (Q2): {percentil50:.2f}")
    print(f"Percentil 75 (Q3): {percentil75:.2f}\n")

    return {
        'media': media,
        'mediana': mediana,
        'desviacion': desviacion_estandar,
        'percentil25': percentil25,
        'percentil50': percentil50,
        'percentil75': percentil75,
    }

resultado_costo = calcular(variable_n1, "Precio Unitario")
resultado_ventas = calcular(variable_n2, "Venta Total")
```

```python
#Normalizar o escalar una variable con NumPy.
#debemos escalar los datos entre 0 y 1, buscar una variable minima y una maxima
costo_min = np.min(variable_n1)
costo_max = np.max(variable_n1)


#aplicamos la fromula (variable - min) / (max - min)
costo_normalizado = (variable_n1 - costo_min) / (costo_max - costo_min)

#verificacion del min y el max
print("Mínimo:", np.min(costo_normalizado)) 
print("Máximo:", np.max(costo_normalizado)) 
```

```python
#La columna 'Costo_Unitario' decidimos rellenar/completarla con el metodo median()
#Buscamos el promedio de costo unitario 
mediana_costo = ventas['Costo_Unitario'].median()
print(mediana_costo)

ventas['Costo_Unitario'] = ventas['Costo_Unitario'].fillna(mediana_costo)
print(ventas['Costo_Unitario'].isnull().sum()) 
```

```python
#Para no perder datos hemos decidido rellenar los valores nulos con una etiqueta

ventas['Metodo_Pago'] = ventas['Metodo_Pago'].fillna('otros')
print(ventas['Metodo_Pago'].isnull().sum())
print(ventas['Metodo_Pago'])
```
